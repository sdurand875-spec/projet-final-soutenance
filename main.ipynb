# Video timing of the Illinois agility test

Entry point of the Python part of the project.
The time of a test (start and finish on the same line) is measured from a
slow-motion video, by manually selecting the START and END frames.
Results are saved in `results/video_measures.csv` and analysed in R (`main.qmd`).

All functions are in `sources/video_timing.py`.

```python
# --- Libraries ---
# Required: opencv-python, numpy, pandas
#   install with:  pip install opencv-python numpy pandas

from pathlib import Path
import pandas as pd

# Functions of the project (sources/video_timing.py)
from sources.video_timing import get_video_info, ask_fps, list_videos, \
                                 load_results, measure_all_videos, anonymize_results
```

## Settings

- `MEASURE = False`: only show the results already measured (no video needed).
- `MEASURE = True`: open the measurement window for the videos not yet measured.

```python
# --- Settings ---

MEASURE = True                      # True = measure videos, False = show results only

video_folder = Path("Video")         # folder with the videos (not shared on GitHub)
results_folder = Path("results")
results_folder.mkdir(exist_ok=True)
results_csv = results_folder / "video_measures.csv"

# Private files (with names, NOT shared on GitHub)
data_folder = Path("data")
data_folder.mkdir(exist_ok=True)
info_csv = data_folder / "children_info.csv"      #name,age,sex,group
id_table_csv = results_folder / "id_table.csv"       # name -> id

# Public file (anonymous, shared on GitHub, used in R)
anonymous_csv = results_folder / "measures_anonymous.csv"
```

## Measurement

Keys: right/left arrow = ±1 frame (hold to scroll), up/down arrow = ±10 frames,
click/drag on the bar = jump. Enter = START, Enter = END, Enter = save.
Backspace = undo, Escape = stop the session.

```python
# --- Measure the videos and anonymize (only if MEASURE = True) ---

if MEASURE:
    first_video = list_videos(video_folder)[0]
    file_fps, _ = get_video_info(first_video)
    real_fps = ask_fps(file_fps)
    measure_all_videos(video_folder, results_csv, real_fps)
    anonymize_results(results_csv, id_table_csv, info_csv, anonymous_csv)
else:
    print("MEASURE = False: measurement skipped, results are read from the CSV.")
```
